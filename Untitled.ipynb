```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error
from pmdarima import auto_arima
from prophet import Prophet
from joblib import dump, load
```

```python
df = pd.read_csv('All_India_Index_july2019_20Aug2020_dec20_1_2.csv')
```

```python
df1 = df[df['Sector']=="Rural"]
```

```python
df2 = df[df['Sector']=="Urban"]
```

```python
df3 = df[df['Sector']=="Rural+Urban"]
```

```python
df1["Date"] = pd.to_datetime(df1["Month"] + ' ' + df1["Year"].astype(str), format='%B %Y')
```

```python
df2["Date"] = pd.to_datetime(df2["Month"] + ' ' + df2["Year"].astype(str), format='%B %Y')
```

```python
df3["Date"] = pd.to_datetime(df3["Month"] + ' ' + df3["Year"].astype(str), format='%B %Y')
```

```python
df1.head()
```

```python
def model_func(df, i):
    df1 = df[["Date",i]]
    df1.set_index("Date", inplace=True)
    df1 = df1.dropna(subset=[i])
    
    plt.figure(figsize=(10, 5))
    plt.plot(df1.index, df1[i], label= f"Simulated {i} Demand")
    plt.title(f"{i} Demand (2015-2023)")
    plt.xlabel("Year")
    plt.ylabel("Demand (Tons)")
    plt.legend()
    plt.grid()
    plt.show()

    train = df1.loc[:"2022-01-01"]
    test = df1.loc["2022-01-01":]

    auto_model = auto_arima(train[i], seasonal=True, m=12, trace=True)
    print(f"Best ARIMA Order: {auto_model.order}")

    model = ARIMA(train[i], order=auto_model.order)  
    fitted_model = model.fit()

    forecast = fitted_model.get_forecast(steps=len(test))
    forecast_values = forecast.predicted_mean
    confidence_intervals = forecast.conf_int()

    plt.figure(figsize=(10, 5))
    plt.plot(train.index, train[i], label="Train Data")
    plt.plot(test.index, test[i], label="Actual Test Data")
    plt.plot(test.index, forecast_values, label="ARIMA Forecast", color="red")
    plt.fill_between(test.index, confidence_intervals.iloc[:, 0], confidence_intervals.iloc[:, 1], color="pink", alpha=0.3)
    plt.title("SARIMA Forecast vs Actual Demand")
    plt.xlabel("Year")
    plt.ylabel(f"{i} Demand (Tons)")
    plt.legend()
    plt.grid()
    plt.show()

    rmse = np.sqrt(mean_squared_error(test[i], forecast_values))
    print(f"RMSE: {rmse:.2f}")
    

    forecast = fitted_model.get_forecast(steps=24)
    predicted = forecast.predicted_mean
    conf_int = forecast.conf_int()

    n_periods = 24
    last_date = df1.index[-1]
    future_dates = pd.date_range(start=last_date + pd.DateOffset(months=1), periods=n_periods, freq='MS')
    predicted.index = future_dates
    conf_int.index = future_dates

    plt.figure(figsize=(10, 5))
    plt.plot(df1, label='Historical')
    plt.plot(predicted.index, predicted, color='orange', label='Forecast')
    plt.fill_between(conf_int.index, conf_int.iloc[:, 0], conf_int.iloc[:, 1], color='orange', alpha=0.3)
    plt.legend()
    plt.title(f'Forecast for {i}')
    plt.grid(True)
    plt.show()

    dump(fitted_model, f'{i}_arima_model.joblib')
```

```python
model_func(df1, "Egg")
```
